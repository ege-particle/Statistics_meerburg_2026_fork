```python
import numpy as np
from scipy import integrate
from matplotlib import pyplot as plt
```

```python
x = np.array([67.4, 73.0, 69.8, 71.2, 68.6, 72.1, 70.4, 74.1, 66.9, 71.7, 69.1, 72.8])
e = np.array([0.7,  1.0,  1.3,  1.8,  0.9,  1.5,  1.1,  2.0,  1.4,  1.2,  0.8,  1.6])
N = len(x)
```

```python
mu_min, mu_max = 60, 80
sigma_min, sigma_max = 0, 10

# Likelihood functions
def likelihood_A(mu):
    # Product of Gaussians: N(x | mu, e^2)
    var = e**2
    logL = -0.5 * np.sum(np.log(2 * np.pi * var) + (x - mu)**2 / var)
    return np.exp(logL)

def likelihood_B(mu, sigma_int):
    # Product of Gaussians: N(x | mu, e^2 + sigma_int^2)
    var = e**2 + sigma_int**2
    logL = -0.5 * np.sum(np.log(2 * np.pi * var) + (x - mu)**2 / var)
    return np.exp(logL)

# Evidence for Model A (1D Integral)
prior_mu = 1.0 / (mu_max - mu_min)
Z_A, _ = integrate.quad(lambda mu: likelihood_A(mu) * prior_mu, mu_min, mu_max)

# Evidence for Model B (2D Integral)
prior_sigma = 1.0 / (sigma_max - sigma_min)
Z_B, _ = integrate.dblquad(lambda s, m: likelihood_B(m, s) * prior_mu * prior_sigma, 
                 mu_min, mu_max, lambda m: sigma_min, lambda m: sigma_max)

odds_ratio = Z_B / Z_A

print(f"Evidence Model A: {Z_A:.2e}")
print(f"Evidence Model B: {Z_B:.2e}")
print(f"Odds Ratio (B/A): {odds_ratio:.2f}")
```
