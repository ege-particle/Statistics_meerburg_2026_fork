```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.stats import norm
from scipy.stats import beta, betabinom
```

```python
n = 20
k = 14
a = 1
b = 1
theta = np.linspace(0, 1, 10000)
prior = beta.pdf(theta, a, b)
L = theta**k * (1-theta)**(n-k)
post_unn = L * prior
post = post_unn / post_unn.sum()
plt.plot(theta, post)
plt.show()
```

```python
mean = np.average(theta, weights=post)
var = np.average((theta-mean)**2, weights=post)
cumulative_posterior = np.cumsum(post)
lower_idx = np.searchsorted(cumulative_posterior, 0.025)
upper_idx = np.searchsorted(cumulative_posterior, 0.975)
ci_low = theta[lower_idx]
ci_high = theta[upper_idx]
next_toss = betabinom.pmf(k+1, n+1, a, b)
more_toss = betabinom.pmf(k+8, n+10, a, b)
print(mean)
print(var)
print(np.array([ci_low, ci_high]))
print(next_toss)
print(more_toss)
```
