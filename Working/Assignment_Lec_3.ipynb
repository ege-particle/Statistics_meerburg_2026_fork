```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 100000
std = 2
xi = np.abs(np.random.normal(loc=0, scale=std, size=N))
func = lambda x: x**3
integral = 0.5 * np.sqrt(2*np.pi*std**2) * np.mean(func(xi))
print(integral)
```

```python
N = 10000
deaths = np.array([0, 1, 2, 3, 4])
corps = np.array([109, 65, 22, 3, 1])
corps_total = 200
probs = corps/corps_total
death_total = np.sum(deaths * corps)
mean = death_total/corps_total
dist = np.random.poisson(lam = mean, size = N)
probs_theory = [np.sum(dist == k) / N for k in deaths]

plt.bar(deaths - 0.1, probs, width = 0.2, label='Observed Data', color='gray')
plt.bar(deaths + 0.1, probs_theory, width=0.2, label=f'Simulated (Poisson, $\lambda$={mean})')
plt.plot(deaths, probs_theory, color = 'red')

plt.xlabel('Number of deaths per corps per year')
plt.ylabel('Probability')
plt.title('Bortkiewicz Horse-Kick Data vs. Poisson Distribution')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
