```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
p_G_G = 0.9
p_G_B = 0.1
p_B_G = 0.5
p_B_B = 0.5
N = 50000

chain = np.zeros(N)
today = 0
for i in range(N):
    chain[i] = today
    p = np.random.rand()
    if today == 1:
        if p < p_G_B:
            today = 0
    else:
        if p < p_B_G:
            today = 1
```

```python
plt.figure(figsize=(12, 4))
plt.plot(chain[:200], 'o-', markersize=3, alpha=0.7)
plt.yticks([0, 1], ['Bad (0)', 'Good (1)'])
plt.xlabel('Day')
plt.title('Trace Plot (First 200 steps)')
plt.grid(axis='y', linestyle='--')
plt.show()
mean_p = np.mean(chain)
print(mean_p)
```

```python
num_blocks = 200
block_size = len(chain) // num_blocks
blocks = [np.mean(chain[i*block_size:(i+1)*block_size]) for i in range(num_blocks)]
uncertainty = np.std(blocks)
plt.hist(blocks)
plt.show()
print(np.mean(blocks))
print(uncertainty)
```

```python
burn_in = 1000
sample = chain[burn_in:]
plt.figure(figsize=(12, 4))
plt.plot(sample[:200], 'o-', markersize=3, alpha=0.7)
plt.yticks([0, 1], ['Bad (0)', 'Good (1)'])
plt.xlabel('Day')
plt.title('Trace Plot (First 200 steps)')
plt.grid(axis='y', linestyle='--')
plt.show()
mean_p = np.mean(sample)
print(mean_p)
```
